- train, test의 age 채우기
      - 위에서 사용했던 Pclass,Sex 컬럼을 그룹핑해서 age의 집계를 확인하고 채워보자

```python
#평균값을 쓸지, 중앙값을 쓸지 판단을 위해 Age컬럼의 기술통계량을 확인
train['Age'].describe()
```

```python
# age는 pclass와 상관관계가 높음
# pclass, sex를 그룹핑해서 age의 평균 출력
gb1 = train[['Pclass','Sex','Age']].groupby(['Pclass','Sex']).mean()
gb1
```

```python
# gb1 - index가 2개 - pclass,sex
#멀티인덱스 
#2등석, 남성 " 퍙군 나이
gb1.loc[(2,'male')]
```

```python
#train Age 결측치가 있는 행 출력
train[train['Age'].isnull()]
```

```python
gb1.loc[(2,'male')]
```

```python
# 규칙에 맞게 Pclass와 Sex에 따른 Age 값 반환하는 함수
def fill_age(row) : 
    # row : 받아오는 행의 데이터(매개변수)
    
    #만약에 Age 컬럼이 결측치하면, gb1 테이블의 값을 인덱싱
    if np.isnan(row['Age']) : 
        return gb1.loc[(row['Pclass'],row['Sex'])]

#결측치가 아니라면 원래 나이 값을 사용
    else : 
        return row['Age']
```

```python
train.apply(fill_age, axis = 1).astype(np.int64)
```

```python
  test['Age'] = test.apply(fill_age, axis = 1).astype(np.int64)
```

- 결과값ㄹ,호 Age를 채울 수 있는 대표값 반환
